```python
import pandas as pd
import re

# =========================================================
# LOAD DATA
# =========================================================

df = pd.read_csv("customers.csv")

print("Original shape:", df.shape)
print(df.head())


# =========================================================
# 1. CUSTOMER ID
# =========================================================

df["customer_id"] = (
    df["customer_id"]
    .astype(str)
    .str.strip()
    .str.upper()
)


# =========================================================
# 2. CUSTOMER NAME
# =========================================================

df["customer_name"] = (
    df["customer_name"]
    .astype(str)
    .str.strip()
    .str.title()
)


# =========================================================
# 3. EMAIL
# =========================================================

df["email"] = (
    df["email"]
    .astype(str)
    .str.strip()
    .str.replace("\\", "", regex=False)
    .str.lower()
)


# =========================================================
# 4. PHONE NUMBER
# =========================================================

def clean_phone(phone):

    phone = str(phone).strip()

    # Keep only numbers
    phone = re.sub(r"\D", "", phone)

    # Remove India country code
    if phone.startswith("91") and len(phone) == 12:
        phone = phone[2:]

    # Remove leading 0
    if phone.startswith("0") and len(phone) == 11:
        phone = phone[1:]

    return phone


df["phone"] = df["phone"].apply(clean_phone)


# =========================================================
# 5. CITY
# =========================================================

df["city"] = (
    df["city"]
    .astype(str)
    .str.strip()
    .str.title()
)


# =========================================================
# 6. STATE
# =========================================================

df["state"] = (
    df["state"]
    .astype(str)
    .str.strip()
    .str.upper()
)


# =========================================================
# 7. PINCODE
# =========================================================

df["pincode"] = (
    df["pincode"]
    .astype(str)
    .str.strip()
)

# Remove .0 if pandas converted it to float
df["pincode"] = df["pincode"].str.replace(
    r"\.0$",
    "",
    regex=True
)


# =========================================================
# 8. SIGNUP DATE
# =========================================================

df["signup_date"] = pd.to_datetime(
    df["signup_date"],
    dayfirst=True,
    errors="coerce"
)

df["signup_date"] = df["signup_date"].dt.strftime("%Y-%m-%d")


# =========================================================
# 9. CUSTOMER SEGMENT
# =========================================================

df["customer_segment"] = (
    df["customer_segment"]
    .astype(str)
    .str.strip()
    .str.lower()
)

segment_mapping = {
    "smb": "SMB",
    "enterprise": "Enterprise",
    "retail": "Retail"
}

df["customer_segment"] = (
    df["customer_segment"]
    .map(segment_mapping)
)


# =========================================================
# 10. PRIME MEMBER
# =========================================================

df["is_prime_member"] = (
    df["is_prime_member"]
    .astype(str)
    .str.strip()
    .str.lower()
)

prime_mapping = {
    "yes": True,
    "y": True,
    "true": True,
    "1": True,

    "no": False,
    "n": False,
    "false": False,
    "0": False
}

df["is_prime_member"] = (
    df["is_prime_member"]
    .map(prime_mapping)
)


# =========================================================
# 11. LIFETIME ORDERS
# =========================================================

df["lifetime_orders"] = pd.to_numeric(
    df["lifetime_orders"],
    errors="coerce"
)

df["lifetime_orders"] = (
    df["lifetime_orders"]
    .fillna(0)
    .astype(int)
)


# =========================================================
# 12. AVERAGE RATING
# =========================================================

df["avg_rating_given"] = pd.to_numeric(
    df["avg_rating_given"],
    errors="coerce"
)

# Ratings should normally be between 0 and 5
df.loc[
    (df["avg_rating_given"] < 0) |
    (df["avg_rating_given"] > 5),
    "avg_rating_given"
] = None


# =========================================================
# 13. REMOVE DUPLICATE CUSTOMERS
# =========================================================

df = df.drop_duplicates(
    subset=["customer_id"],
    keep="first"
)


# =========================================================
# 14. CHECK MISSING VALUES
# =========================================================

print("\nMissing values:")
print(df.isnull().sum())


# =========================================================
# 15. CHECK DUPLICATES
# =========================================================

print("\nDuplicate customer IDs:")
print(df["customer_id"].duplicated().sum())


# =========================================================
# 16. FINAL COLUMN ORDER
# =========================================================

columns = [
    "customer_id",
    "customer_name",
    "email",
    "phone",
    "city",
    "state",
    "pincode",
    "signup_date",
    "customer_segment",
    "is_prime_member",
    "lifetime_orders",
    "avg_rating_given"
]

df = df[columns]


# =========================================================
# 17. SAVE CLEANED DATA
# =========================================================

df.to_csv(
    "customer_cleaned.csv",
    index=False
)


# =========================================================
# 18. DISPLAY RESULT
# =========================================================

print("\nCleaned shape:", df.shape)

print("\nCleaned customer data:")
print(df)
```

```python
# ============================================================
# SMARTLOGIX - DELIVERY LOG DATA CLEANING
# ============================================================

import pandas as pd
import numpy as np


# ============================================================
# 1. LOAD DATASET
# ============================================================

input_file = "delivery_logs.csv"
output_file = "delivery_logs_cleaned.csv"

df = pd.read_csv(
    input_file,
    dtype=str,
    keep_default_na=True
)

print("=" * 60)
print("DELIVERY LOG DATA CLEANING")
print("=" * 60)

print("\nOriginal shape:")
print(df.shape)

print("\nOriginal columns:")
print(df.columns.tolist())


# ============================================================
# 2. CLEAN COLUMN NAMES
# ============================================================

df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print("\nCleaned columns:")
print(df.columns.tolist())


# ============================================================
# 3. REMOVE EXTRA SPACES FROM ALL STRING COLUMNS
# ============================================================

for col in df.columns:

    if df[col].dtype == "object":

        df[col] = df[col].apply(
            lambda x: x.strip()
            if isinstance(x, str)
            else x
        )


# ============================================================
# 4. CLEAN log_id
# ============================================================

df["log_id"] = (
    df["log_id"]
    .astype("string")
    .str.strip()
    .str.upper()
)


# ============================================================
# 5. CLEAN order_id
# ============================================================

df["order_id"] = (
    df["order_id"]
    .astype("string")
    .str.strip()
    .str.upper()
)


# ============================================================
# 6. CLEAN event_seq
# ============================================================

df["event_seq"] = pd.to_numeric(
    df["event_seq"],
    errors="coerce"
)

df["event_seq"] = (
    df["event_seq"]
    .astype("Int64")
)


# ============================================================
# 7. CLEAN event_type
# ============================================================

df["event_type"] = (
    df["event_type"]
    .astype("string")
    .str.strip()
    .str.upper()
)


# ============================================================
# 8. CLEAN event_timestamp
# ============================================================

def clean_timestamp(value):

    # Missing value
    if pd.isna(value):
        return pd.NaT

    value = str(value).strip()

    # Empty value
    if value == "":
        return pd.NaT

    # --------------------------------------------------------
    # Unix timestamp
    # --------------------------------------------------------

    if value.isdigit():

        number = int(value)

        # 10 digits = Unix seconds
        if len(value) == 10:

            return pd.to_datetime(
                number,
                unit="s",
                errors="coerce"
            )

        # 13 digits = Unix milliseconds
        elif len(value) == 13:

            return pd.to_datetime(
                number,
                unit="ms",
                errors="coerce"
            )

    # --------------------------------------------------------
    # Normal / ISO datetime
    # --------------------------------------------------------

    return pd.to_datetime(
        value,
        errors="coerce"
    )


# Apply timestamp cleaning
df["event_timestamp"] = (
    df["event_timestamp"]
    .apply(clean_timestamp)
)


# IMPORTANT:
# Force the complete column to datetime
df["event_timestamp"] = pd.to_datetime(
    df["event_timestamp"],
    errors="coerce"
)


# Check datatype
print("\nEvent timestamp datatype:")
print(df["event_timestamp"].dtype)


# ============================================================
# 9. CLEAN hub_code
# ============================================================

df["hub_code"] = (
    df["hub_code"]
    .astype("string")
    .str.strip()
    .str.upper()
)


# ============================================================
# 10. CLEAN vehicle_id
# ============================================================

df["vehicle_id"] = (
    df["vehicle_id"]
    .astype("string")
    .str.strip()
    .str.upper()
)


# ============================================================
# 11. CLEAN scanned_by_emp
# ============================================================

df["scanned_by_emp"] = (
    df["scanned_by_emp"]
    .astype("string")
    .str.strip()
    .str.upper()
)


# ============================================================
# 12. CLEAN location_city
# ============================================================

df["location_city"] = (
    df["location_city"]
    .astype("string")
    .str.strip()
    .str.title()
)


# ============================================================
# 13. CLEAN remarks
# ============================================================

df["remarks"] = (
    df["remarks"]
    .astype("string")
)


# Replace newlines / tabs with spaces
df["remarks"] = (
    df["remarks"]
    .str.replace(
        r"[\r\n\t]+",
        " ",
        regex=True
    )
)


# Remove multiple spaces
df["remarks"] = (
    df["remarks"]
    .str.replace(
        r"\s+",
        " ",
        regex=True
    )
)


# Remove leading/trailing spaces
df["remarks"] = (
    df["remarks"]
    .str.strip()
)


# Convert empty remarks to missing value
df["remarks"] = (
    df["remarks"]
    .replace(
        ["", "nan", "None", "<NA>"],
        pd.NA
    )
)


# ============================================================
# 14. CLEAN exception_code
# ============================================================

df["exception_code"] = (
    df["exception_code"]
    .astype("string")
    .str.strip()
    .str.upper()
)


# Missing exception code → NA
df["exception_code"] = (
    df["exception_code"]
    .replace(
        ["", "NAN", "NONE", "<NA>"],
        "NA"
    )
)


# ============================================================
# 15. CHECK DUPLICATE LOG IDs
# ============================================================

duplicate_count = (
    df["log_id"]
    .duplicated()
    .sum()
)

print("\nDuplicate log IDs:")
print(duplicate_count)


# ============================================================
# 16. REMOVE DUPLICATE LOG IDs
# ============================================================

df = df.drop_duplicates(
    subset=["log_id"],
    keep="first"
)


# ============================================================
# 17. CHECK DUPLICATE ORDER + EVENT SEQUENCE
# ============================================================

duplicate_events = df[
    df.duplicated(
        subset=[
            "order_id",
            "event_seq"
        ],
        keep=False
    )
]

print("\nDuplicate order_id + event_seq combinations:")
print(len(duplicate_events))


# ============================================================
# 18. CHECK MISSING VALUES
# ============================================================

print("\n" + "=" * 60)
print("MISSING VALUE CHECK")
print("=" * 60)

missing_values = df.isnull().sum()

print(missing_values)


# ============================================================
# 19. CHECK INVALID TIMESTAMPS
# ============================================================

invalid_dates = df[
    df["event_timestamp"].isna()
]

print("\n" + "=" * 60)
print("INVALID TIMESTAMP CHECK")
print("=" * 60)

print("Invalid timestamp rows:", len(invalid_dates))

if len(invalid_dates) > 0:

    print(
        invalid_dates[
            [
                "log_id",
                "order_id",
                "event_timestamp"
            ]
        ]
    )


# ============================================================
# 20. CHECK EVENT TYPES
# ============================================================

print("\n" + "=" * 60)
print("EVENT TYPE CHECK")
print("=" * 60)

print(
    df["event_type"]
    .value_counts(dropna=False)
)


# ============================================================
# 21. CHECK HUB CODES
# ============================================================

print("\n" + "=" * 60)
print("HUB CODE CHECK")
print("=" * 60)

print(
    df["hub_code"]
    .value_counts(dropna=False)
)


# ============================================================
# 22. CHECK LOCATION CITIES
# ============================================================

print("\n" + "=" * 60)
print("LOCATION CITY CHECK")
print("=" * 60)

print(
    df["location_city"]
    .value_counts(dropna=False)
)


# ============================================================
# 23. CHECK EXCEPTION CODES
# ============================================================

print("\n" + "=" * 60)
print("EXCEPTION CODE CHECK")
print("=" * 60)

print(
    df["exception_code"]
    .value_counts(dropna=False)
)


# ============================================================
# 24. VALIDATE EVENT SEQUENCE
# ============================================================

invalid_event_seq = df[
    df["event_seq"].isna() |
    (df["event_seq"] <= 0)
]

print("\n" + "=" * 60)
print("EVENT SEQUENCE CHECK")
print("=" * 60)

print(
    "Invalid event sequence rows:",
    len(invalid_event_seq)
)


# ============================================================
# 25. SORT DELIVERY LOG
# ============================================================

df = df.sort_values(
    by=[
        "order_id",
        "event_seq",
        "event_timestamp"
    ],
    ascending=[
        True,
        True,
        True
    ]
)


# ============================================================
# 26. RESET INDEX
# ============================================================

df = df.reset_index(drop=True)


# ============================================================
# 27. FINAL COLUMN ORDER
# ============================================================

final_columns = [

    "log_id",
    "order_id",
    "event_seq",
    "event_type",
    "event_timestamp",
    "hub_code",
    "vehicle_id",
    "scanned_by_emp",
    "location_city",
    "remarks",
    "exception_code"

]

df = df[final_columns]


# ============================================================
# 28. FORMAT TIMESTAMP
# ============================================================

df["event_timestamp"] = (
    df["event_timestamp"]
    .dt.strftime(
        "%Y-%m-%d %H:%M:%S"
    )
)


# ============================================================
# 29. SAVE CLEANED DATASET
# ============================================================

df.to_csv(
    output_file,
    index=False
)


# ============================================================
# 30. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("CLEANING COMPLETED SUCCESSFULLY")
print("=" * 60)

print("\nOriginal rows:")
print(len(pd.read_csv(input_file)))

print("\nFinal rows:")
print(len(df))

print("\nFinal columns:")
print(df.columns.tolist())

print("\nOutput file:")
print(output_file)

print("\nFinal dataset preview:")
print(df.head(10))

print("\nFinal data types:")
print(df.dtypes)

print("\n" + "=" * 60)
print("DONE")
print("=" * 60)
```
