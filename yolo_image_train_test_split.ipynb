```python
# ============================================================
# DRONE DATASET PREPARATION
# Classification Dataset
# ============================================================

import os
import shutil
import random
from pathlib import Path

# ============================================================
# 1. PATHS
# ============================================================

SOURCE_DIR = Path(
    r"C:\ML Project\SmartLogix AI_Intelligent_Multi_Modal_Logistics\smartlogix drone images"
)

OUTPUT_DIR = Path(
    r"C:\ML Project\SmartLogix AI_Intelligent_Multi_Modal_Logistics\drone_classification"
)

# ============================================================
# 2. CLASS FOLDER MAPPING
# ============================================================

CLASS_MAPPING = {
    "broken propeller": "broken_propeller",
    "frame damage": "frame_damage",
    "motor damage": "motor_damage",
    "landing gear": "landing_gear_damage",
    "battery damage": "battery_damage"
}

# ============================================================
# 3. SPLIT RATIO
# ============================================================

TRAIN_RATIO = 0.70
VAL_RATIO = 0.20
TEST_RATIO = 0.10

# ============================================================
# 4. RANDOM SEED
# ============================================================

random.seed(42)

# ============================================================
# 5. IMAGE EXTENSIONS
# ============================================================

IMAGE_EXTENSIONS = {
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".webp"
}

# ============================================================
# 6. CHECK SOURCE DIRECTORY
# ============================================================

if not SOURCE_DIR.exists():
    raise FileNotFoundError(
        f"Source folder not found:\n{SOURCE_DIR}"
    )

# ============================================================
# 7. CREATE OUTPUT DIRECTORIES
# ============================================================

for split in ["train", "val", "test"]:

    for class_name in CLASS_MAPPING.values():

        folder = OUTPUT_DIR / split / class_name

        folder.mkdir(
            parents=True,
            exist_ok=True
        )

# ============================================================
# 8. PROCESS EACH CLASS
# ============================================================

for source_folder_name, class_name in CLASS_MAPPING.items():

    source_folder = SOURCE_DIR / source_folder_name

    print("\n==========================================")
    print("Processing:", source_folder_name)
    print("Class:", class_name)
    print("==========================================")

    if not source_folder.exists():

        print(
            f"WARNING: Folder not found: {source_folder}"
        )

        continue

    # --------------------------------------------------------
    # Find images
    # --------------------------------------------------------

    images = [
        file
        for file in source_folder.iterdir()
        if file.is_file()
        and file.suffix.lower() in IMAGE_EXTENSIONS
    ]

    print("Total images:", len(images))

    if len(images) == 0:

        print("WARNING: No images found.")

        continue

    # --------------------------------------------------------
    # Shuffle
    # --------------------------------------------------------

    random.shuffle(images)

    # --------------------------------------------------------
    # Calculate split sizes
    # --------------------------------------------------------

    total = len(images)

    train_count = int(total * TRAIN_RATIO)

    val_count = int(total * VAL_RATIO)

    # Remaining images go to test
    test_count = total - train_count - val_count

    # --------------------------------------------------------
    # Split
    # --------------------------------------------------------

    train_images = images[:train_count]

    val_images = images[
        train_count:
        train_count + val_count
    ]

    test_images = images[
        train_count + val_count:
    ]

    print("Train:", len(train_images))
    print("Validation:", len(val_images))
    print("Test:", len(test_images))

    # --------------------------------------------------------
    # Copy function
    # --------------------------------------------------------

    def copy_images(image_list, split_name):

        destination = (
            OUTPUT_DIR
            / split_name
            / class_name
        )

        for index, image_path in enumerate(image_list):

            # Create unique filename
            new_name = (
                f"{class_name}_{index:05d}"
                f"{image_path.suffix.lower()}"
            )

            destination_path = (
                destination / new_name
            )

            shutil.copy2(
                image_path,
                destination_path
            )

    # --------------------------------------------------------
    # Copy images
    # --------------------------------------------------------

    copy_images(
        train_images,
        "train"
    )

    copy_images(
        val_images,
        "val"
    )

    copy_images(
        test_images,
        "test"
    )

# ============================================================
# 9. FINAL MESSAGE
# ============================================================

print("\n")
print("=" * 60)
print("DATASET PREPARATION COMPLETED")
print("=" * 60)

print("\nDataset created at:")

print(OUTPUT_DIR)

print("\nStructure:")

print("""
drone_classification/
│
├── train/
│   ├── broken_propeller/
│   ├── frame_damage/
│   ├── motor_damage/
│   ├── landing_gear_damage/
│   └── battery_damage/
│
├── val/
│   ├── broken_propeller/
│   ├── frame_damage/
│   ├── motor_damage/
│   ├── landing_gear_damage/
│   └── battery_damage/
│
└── test/
    ├── broken_propeller/
    ├── frame_damage/
    ├── motor_damage/
    ├── landing_gear_damage/
    └── battery_damage/
""")
```
